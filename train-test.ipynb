```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv(r"C:\Users\HP\Downloads\qml_8features_dataset.csv") 
df.head()
```

```python
feature_cols = [
    "HER2pos", "MRI_pct_change", "MRI_shrinkage_early",
    "is_TN", "PgRpos", "ERpos", "MRI_LD_Baseline", "age",
]
target_col = "PCR"

X = df[feature_cols].values
y = df[target_col].values

print(f"Загалом: {len(df)} пацієнтів, {len(feature_cols)} ознак")
print(f"Баланс класів: No pCR={int((y==0).sum())}, pCR={int((y==1).sum())}")
```

```python
RANDOM_SEED = 42  # той самий seed, що і в команди — не змінюй!

X_train, X_test, y_train, y_test, id_train, id_test = train_test_split(
    X, y, df["SUBJECTID"].values,
    test_size=0.2,
    random_state=RANDOM_SEED,
    stratify=y,
)

print(f"Train: {len(X_train)} пацієнтів (pCR={int(y_train.sum())})")
print(f"Test:  {len(X_test)} пацієнтів (pCR={int(y_test.sum())})")
```

```python
scaler = MinMaxScaler(feature_range=(0, np.pi))

X_train_scaled = scaler.fit_transform(X_train)   
X_test_scaled = scaler.transform(X_test)        

print("Діапазон train після масштабування:", X_train_scaled.min(), "-", X_train_scaled.max())
```
